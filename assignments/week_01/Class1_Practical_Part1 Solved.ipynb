# Practical 7 - Laboratory Temperature Converter

## Problem

A microbiology laboratory records temperatures in Celsius, but an international system requires Fahrenheit.

Ask the user to enter the temperature in Celsius.

Display:
- Celsius
- Fahrenheit (up to 2 decimal places)

## Formula

$$F=(C\times\frac{9}{5})+32$$

### Pseudocode

```
START
Input temperature in Celsius
Convert into Fahrenheit
Display both temperatures
END
```


````python
from traitlets import ForwardDeclaredInstance
# temperature converter
temperature_input = "Celsius"
celsius = 42.5
fahrenheit = (celsius*9/5)+32
print(f'{temperature_input} : {celsius}°C')
print(f'Fahrenheit : {fahrenheit:.2f}°F')
````

# Practical 9 - Student Percentage Calculator

## Problem

Ask the user to enter marks obtained in the following five subjects:

- English
- Mathematics
- Physics
- Chemistry
- Computer Science

Display:
- Total Marks
- Percentage
- Pass/Fail (Pass ≥ 50%) using comparison operators.

## Formula

$$\mathrm{Total}=m_1+m_2+m_3+m_4+m_5$$

$$\mathrm{Percentage}=\frac{\mathrm{Total}}{500}\times100$$

### Pseudocode

```
START
Input marks of five subjects
Calculate Total
Calculate Percentage
Determine PASS/FAIL
Display Total
Display Percentage
Display Result
END
```


````python
# persentage calculatur
# Total subject = 5
eng = 85
math = 90
phy = 75
chem = 70
cs = 95
total = eng+math+phy+chem+cs
persentage = (total/500)*100
print(f'English           : {eng:^15}')
print(f'Math              : {math:^15}')
print(f'Physics           : {phy:^15}')
print(f'Chemistry         : {chem:^15}')
print(f'Computer Science  : {cs:^15}')
print(f'Total             : {total:^15}')
print(f'Persentage        : {persentage:>9}%')
print(f'Result            : {persentage>50:^15}')




````

# Practical 10 - Retail Store Sales Summary

## Problem

A retail store recorded sales for three consecutive days.

| Day | Sales (PKR) |
|---|---:|
| Monday | 18500 |
| Tuesday | 22350 |
| Wednesday | 19800 |

Store the sales values.

Calculate and display:
- Total Sales
- Average Daily Sales

## Formula

$$\mathrm{Total}=Day_1+Day_2+Day_3$$

$$\mathrm{Average}=\frac{\mathrm{Total}}{3}$$

### Pseudocode

```
START
Store sales of three days
Calculate total sales
Calculate average sales
Display total
Display average
END
```


````python
# sale summery
monday = 18500
tuesday = 22350
wednesday = 19800
total = monday+tuesday+wednesday
avg = total/3
print(f'Monday Sales    : {monday:^15}')
print(f'Tuesday Sales   : {tuesday:^15}')
print(f'Wednesday Sales : {wednesday:^15}')
print(f'Total Sales     : {total:^15}')
print(f'Average Sales   : {avg:^15.2f}')
````
