# Question 27 - Strict Email Format Check Using Anchors

## Problem

An analytics pipeline needs to ensure that extracted strings match an exact data format without trailing or leading junk characters.
The system receives three sample records:

| Record | Value |
|---|---|
| record1 | "user@company.com" |
| record2 |   "user@company.com" |
| record3 | "user@company.com extra_text" |

Using `re.search()` along with regex anchors (`^` for start of string and `$` for end of string), write a pattern to strictly validate a standard string format. The pattern should only match if the string begins and ends exactly with the characters matching `\w+@\w+\.\w+`.

Display:
- Whether each record is strictly valid (True/False).

### Pseudocode
```
START
Import re
Store the three records
Define a regex pattern using ^ and $ anchors
Search each record using the pattern
Display whether each record is strictly valid
END
```

````python
import re
record1 = "user@company.com"
record2 = "user@company.com"
record3 = "user@company.com extra_text"
print(f'Record1 : {record1}')
print(f'Record2 : {record2}')
print(f'Record3 : {record3}')
print(f'Record1 : {re.search(r"^\w+@\w+\.\w+$",record1)}')
print(f'Record2 : {re.search(r"^\w+@\w+\.\w+$",record2)}')
print(f'Record3 : {re.search(r"^\w+@\w+\.\w+$",record3)}')
````
